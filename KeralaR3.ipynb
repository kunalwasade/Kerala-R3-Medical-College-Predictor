Import Libraries

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

from catboost import CatBoostClassifier
```

Load Dataset

```python
df = pd.read_excel("C:\\Data Science\\Debug\\ML\\ML PRO\\Kerala R3\\KERALA R3 ALLOTT.xlsx")
```

```python
df.head()
```

Clean Column Names

```python
df.columns = (
    df.columns
    .str.replace("\n", " ", regex=False)
    .str.strip()
)

print(df.columns)
```

Check Dataset

Remove Duplicates

```python
df.drop_duplicates(inplace=True)

print(df.shape)
```

Saparate Features and Target

```python
X = df[
    [
        "Rank",
        "College Type",
        "Course",
        "Candidate Category",
        "Alloted Category"
    ]
]

y = df["College Name"]
```

Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

Categorical Feature Index

```python
cat_features = [1,2,3,4]
```

Create CatBoost Model

```python
model = CatBoostClassifier(

    iterations=1000,
    learning_rate=0.05,
    depth=8,
    loss_function="MultiClass",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=100
)
```

Train Model

```python
model.fit(

    X_train,
    y_train,
    cat_features=cat_features

)
```

Prediction

```python
y_pred = model.predict(X_test)
```

Accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
```

Classification Report

```python
print(classification_report(y_test,y_pred))
```

```python
import pandas as pd

# -------------------------
# User Input
# -------------------------

rank = float(input("Enter Your Rank: "))
college_type = input("Enter College Type: ").strip()
course = input("Enter Course: ").strip()
candidate_category = input("Enter Candidate Category: ").strip()
alloted_category = input("Enter Alloted Category: ").strip()

# -------------------------
# Filter Dataset
# -------------------------

filtered = df[
    (df["College Type"] == college_type) &
    (df["Course"] == course) &
    (df["Candidate Category"] == candidate_category) &
    (df["Alloted Category"] == alloted_category)
].copy()

if filtered.empty:
    print("❌ No matching records found.")

else:

    # -------------------------
    # Calculate Gap
    # -------------------------

    filtered["Gap"] = filtered["Rank"] - rank

    # -------------------------
    # Find Best Match
    # -------------------------

    closest_record = filtered.loc[filtered["Gap"].abs().idxmin()]

    best_match_college = closest_record["College Name"]
    best_match_rank = int(closest_record["Rank"])

    # -------------------------
    # Chance Function
    # -------------------------

    def chance(cutoff_rank, user_rank):

        diff = cutoff_rank - user_rank

        # Previous year cutoff is much higher than your rank
        if diff >= 200:
         return "🟢 High Chance"

       # Previous year cutoff is close to your rank
        elif diff >= -100:
         return "🟡 Borderline"

       # Previous year cutoff is much lower than your rank
        else:
         return "🔴 Tough Chance"

    filtered["Chance"] = filtered["Rank"].apply(
        lambda x: chance(x, rank)
    )

    # -------------------------
    # Keep One Record Per College
    # -------------------------

    result = (
        filtered
        .sort_values("Rank", ascending=False)
        .drop_duplicates(subset="College Name", keep="first")
    )

    # -------------------------
    # Sort by Rank
    # -------------------------

    result = (
        result
        .sort_values("Rank")
        .reset_index(drop=True)
    )

    # -------------------------
    # Top 10 Colleges
    # -------------------------

    #result = result.head(10)

    # -------------------------
    # Serial Number
    # -------------------------

    result.insert(0, "S.No", range(1, len(result)+1))

    # -------------------------
    # Final Columns
    # -------------------------

    result = result[
        [
            "S.No",
            "Course",
            "College Type",
            "Candidate Category",
            "Alloted Category",
            "Rank",
            "College Name",
            "Chance"
        ]
    ]

    # -------------------------
    # Display Settings
    # -------------------------

    pd.set_option("display.max_columns", None)
    pd.set_option("display.max_colwidth", None)
    pd.set_option("display.width", 2000)

    # -------------------------
    # Prediction Summary
    # -------------------------

    print("\n" + "=" * 100)
    print("                        COLLEGE PREDICTION SUMMARY")
    print("=" * 100)

    print(f"Your Rank                  : {int(rank)}")
    print(f"Course                     : {course}")
    print(f"College Type               : {college_type}")
    print(f"Candidate Category         : {candidate_category}")
    print(f"Alloted Category           : {alloted_category}")
    print(f"Best Match College         : {best_match_college}")
    print(f"Previous Year Cutoff Rank  : {best_match_rank}")

    print("=" * 100)

    print("\nTop Recommended Colleges:\n")

    display(
        result.style
        .hide(axis="index")
        .set_properties(
            subset=[
                "Course",
                "College Type",
                "Candidate Category",
                "Alloted Category",
                "College Name",
                "Chance"
            ],
            **{"text-align": "left"}
        )
        .set_properties(
            subset=["S.No", "Rank"],
            **{"text-align": "center"}
        )
        .set_table_styles([
            {
                "selector": "th",
                "props": [
                    ("text-align", "center"),
                    ("font-weight", "bold")
                ]
            }
        ])
    )
```
